```python
import numpy as np
import glob
```

```python
# バブルでーたのPaht取得
bubble_path = glob.glob("/home/cygnus/fujimoto/Cygnus-X_Molecular_Cloud_Analysis/Binary_classification/data/FUGIN/processed_data/*/*/*/bubble_data.npy")
bubble_path = sorted(bubble_path)
```

```python
# バブルデータのカウント
for i in range(len(bubble_path)):
    name = bubble_path[i]
    data_num  = len(np.load(name))
    data_name = name.split("/")
    data_name = data_name[-4] + " " + data_name[-2]
    print(f"{data_name}: {data_num}")
```

```python
# 非バブルでーた（上限、下限）のPaht取得
non_bubble_path = glob.glob("/home/cygnus/fujimoto/Cygnus-X_Molecular_Cloud_Analysis/Binary_classification/data/FUGIN/processed_data/**/thresh_up_low/lower-1.66_data.npy")
non_bubble_path = sorted(non_bubble_path)
```

```python
# バブルデータのカウント
for i in range(len(non_bubble_path)):
    name = non_bubble_path[i]
    data_num  = len(np.load(name))
    data_name = name.split("/")
    data_name = data_name[-3]
    print(f"{data_name}: {data_num}")
```
